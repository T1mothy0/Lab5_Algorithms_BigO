```python
LAST_NAME = "MEJOS"
STUDENT_ID = "TUPM-26-3605"
FAVORITE_ARTIST = "TAYLOR SWIFT"

SEED_DIGIT = int(STUDENT_ID[-1])                              
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())        
NAME_LENGTH = len(LAST_NAME)                                   
ARTIST_LENGTH = len(FAVORITE_ARTIST)                           
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST if c.isalpha()) 

dept_a = [SEED_DIGIT + i * 4 for i in range(5)]
dept_b = [NAME_LENGTH + 3 + i * (ARTIST_LENGTH // 2) for i in range(5)]

print("Department A:", dept_a)
print("Department B:", dept_b)

# Consolidate (merge step)
merged, i, j, comparisons = [], 0, 0, 0
while i < len(dept_a) and j < len(dept_b):
    comparisons += 1
    if dept_a[i] <= dept_b[j]:
        merged.append(dept_a[i])
        i += 1
    else:
        merged.append(dept_b[j])
        j += 1
merged.extend(dept_a[i:])
merged.extend(dept_b[j:])

print("Consolidated:", merged)
print("Records before:", len(dept_a) + len(dept_b), "| Records after:", len(merged))
print("Comparisons:", comparisons)
print("Correctly ordered:", merged == sorted(merged))

sizes = [10, 100, 1000, 10000]

for n in sizes:
    a = [SEED_DIGIT + i * 4 for i in range(n)]
    b = [NAME_LENGTH + 3 + i * (ARTIST_LENGTH // 2) for i in range(n)]

    merged, i, j, comparisons = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])

    print(f"n={n:6d} | Comparisons={comparisons:6d} | Output size={len(merged):6d}")

print("=== FINAL OUTPUT ===")
print(f"Department A ({len(dept_a)} records) and Department B ({len(dept_b)} records)")
print(f"were consolidated into {len(merged)} ordered enrollment records using {comparisons} comparisons.")
print(f"Final order correctly ascending: {merged == sorted(merged)}")
```
