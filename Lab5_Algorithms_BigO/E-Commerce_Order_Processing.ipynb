```python
LAST_NAME = "MEJOS"
STUDENT_ID = "TUPM-26-3605"
FAVORITE_ARTIST = "TAYLOR SWIFT"

SEED_DIGIT = int(STUDENT_ID[-1])                              
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())        
NAME_LENGTH = len(LAST_NAME)                                   
ARTIST_LENGTH = len(FAVORITE_ARTIST)                           
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST if c.isalpha()) 

import time

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

def measure(sort_fn, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_fn(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

# Orders generated from student-specific inputs
orders = [(ID_SUM * 3 + i * 7) % 100 for i in range(8)]
sorted_in = sorted(orders)
reverse_in = sorted(orders, reverse=True)
random_in = orders

print("Generated Order Dataset:", orders)
print("Sorted Input:", sorted_in)
print("Reverse Input:", reverse_in)
print("Random Input:", random_in)

for label, data in [("Sorted", sorted_in), ("Reverse", reverse_in), ("Random", random_in)]:
    print(f"\n--- {label} Input ---")
    for name, fn in [("Bubble", bubble_sort), ("Selection", selection_sort), ("Insertion", insertion_sort)]:
        (result, comp, mov), elapsed = measure(fn, data)
        print(f"{name:10s} | comparisons={comp:3d} | movements={mov:3d} | time={elapsed:.8f}s | sorted correctly={result == sorted(data)}")

sizes = [10, 50, 100, 200]

for n in sizes:
    data = [(ID_SUM * 3 + i * 7) % 1000 for i in range(n)]
    rev = sorted(data, reverse=True)

    _, bc, bs = bubble_sort(rev)
    _, sc, ss = selection_sort(rev)
    _, ic, ishifts = insertion_sort(rev)

    print(f"n={n:4d} | Bubble comp={bc:5d} | Selection comp={sc:5d} | Insertion comp={ic:5d}")

# Movement counts (already computed above — isolating them here)
print("=== MOVEMENT COUNTS ===")
for label, data in [("Sorted", sorted_in), ("Reverse", reverse_in), ("Random", random_in)]:
    _, bc, bs = bubble_sort(data)
    _, sc, ss = selection_sort(data)
    _, ic, ishifts = insertion_sort(data)
    print(f"{label:8s} | Bubble swaps={bs:3d} | Selection swaps={ss:3d} | Insertion shifts={ishifts:3d}")

# Execution times
print("\n=== EXECUTION TIMES ===")
for label, data in [("Sorted", sorted_in), ("Reverse", reverse_in), ("Random", random_in)]:
    _, bt = measure(bubble_sort, data)
    _, st = measure(selection_sort, data)
    _, it = measure(insertion_sort, data)
    print(f"{label:8s} | Bubble={bt:.8f}s | Selection={st:.8f}s | Insertion={it:.8f}s")

# Sorting verification
print("\n=== SORTING VERIFICATION ===")
for label, data in [("Sorted", sorted_in), ("Reverse", reverse_in), ("Random", random_in)]:
    bres, _, _ = bubble_sort(data)
    sres, _, _ = selection_sort(data)
    ires, _, _ = insertion_sort(data)
    correct = bres == sorted(data) and sres == sorted(data) and ires == sorted(data)
    print(f"{label:8s} | All three correctly sorted: {correct}")

# Big O analysis (written conclusion, printed as a label)
print("\n=== BIG O ANALYSIS ===")
print("Bubble/Insertion: O(n) best case (sorted), O(n^2) worst case (reverse)")
print("Selection: O(n^2) in every case (no early exit)")

# Best-, average-, worst-case observations
print("\n=== CASE OBSERVATIONS ===")
print("Best case  -> Sorted input   (fewest comparisons/movements for Bubble & Insertion)")
print("Average    -> Random input   (moderate comparisons/movements)")
print("Worst case -> Reverse input  (maximum comparisons/movements for all three)")

# Final Output
print("\n=== FINAL OUTPUT ===")
print("All sorting algorithms correctly produced ascending order across all three input conditions.")
print("Selection Sort's comparison count stayed fixed at 28 regardless of input order (no early exit),")
print("while Bubble and Insertion Sort dropped to their minimum on sorted input and peaked on reverse input,")
print("directly demonstrating the best/average/worst-case distinction.")
```
