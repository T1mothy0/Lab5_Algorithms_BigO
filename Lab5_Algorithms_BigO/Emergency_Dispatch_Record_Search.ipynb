```python
LAST_NAME = "MEJOS"
STUDENT_ID = "TUPM-26-3605"
FAVORITE_ARTIST = "TAYLOR SWIFT"

SEED_DIGIT = int(STUDENT_ID[-1])                              
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())        
NAME_LENGTH = len(LAST_NAME)                                   
ARTIST_LENGTH = len(FAVORITE_ARTIST)                           
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST if c.isalpha()) 

# Incident records — unique values derived from student inputs
original = [NAME_LENGTH * 10 + i * 5 for i in range(10)]

# Rotation point from SEED_DIGIT
rp = SEED_DIGIT % len(original)
rotated = original[rp:] + original[:rp]

# Target incident from ID_SUM and ARTIST_SUM
target = original[(ID_SUM + ARTIST_SUM) % len(original)]

print("Original Dataset:", original)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)

# Linear Search (Requirement 5: compare with linear search)
lc = 0
li = -1
for i in range(len(rotated)):
    lc += 1
    if rotated[i] == target:
        li = i
        break
if li == -1:
    print("Linear Search: NOT FOUND |", "comparisons:", lc)
else:
    print(f"Linear Search -> position={li}, comparisons={lc}")

# Modified Binary Search for a rotated sorted array
low, high, rc = 0, len(rotated) - 1, 0
ri = -1
while low <= high:
    mid = (low + high) // 2
    rc += 1
    if rotated[mid] == target:
        ri = mid
        break
    if rotated[low] <= rotated[mid]:
        if rotated[low] <= target < rotated[mid]:
            high = mid - 1
        else:
            low = mid + 1
    else:
        if rotated[mid] < target <= rotated[high]:
            low = mid + 1
        else:
            high = mid - 1
if ri == -1:
    print("Rotated Search: NOT FOUND |", "comparisons:", rc)
else:
    print(f"Rotated Search -> position={ri}, comparisons={rc}")

sizes = [10, 100, 1000, 10000]

for n in sizes:
    data = [NAME_LENGTH * 10 + i * 5 for i in range(n)]
    rp = SEED_DIGIT % n
    rot = data[rp:] + data[:rp]
    tgt = data[(ID_SUM + ARTIST_SUM) % n]

    lc = 0
    for v in rot:
        lc += 1
        if v == tgt:
            break

    low, high, rc = 0, n - 1, 0
    while low <= high:
        mid = (low + high) // 2
        rc += 1
        if rot[mid] == tgt:
            break
        if rot[low] <= rot[mid]:
            if rot[low] <= tgt < rot[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if rot[mid] < tgt <= rot[high]:
                low = mid + 1
            else:
                high = mid - 1

    print(f"n={n:6d} | Linear={lc:6d} | Rotated={rc:3d}")   

print("=== FINAL OUTPUT ===")
print(f"Target incident {target} was located at position {ri} in the rotated dataset")
print(f"using {rc} comparisons (modified binary search),")
print(f"compared to {lc} comparisons required by linear search on the same data,")
print(f"a reduction of {lc - rc} comparisons.")
```
